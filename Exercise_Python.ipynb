```python
# Mirron image
n = 5
for i in range(1,2*n):
  if i <= n:
    star=i
  else:
    star = 2*n-i
  space = n-star
  for j in range(0,space):
    print(" ",end=" ")
  for k in range(0,2*star-1):
    print("* ",end="")

  print()
```

```python
# [ Instructor name , Duration ] python course
course = {
    "java":{"instructor":"priya","duration":32},
    "sql":{"instructor":"riya","duration":32},
    "python":{"instructor":"Abhisake","duration":42},
    "php":{"instructor":"naman","duration":32}
}
for course,details in course.items():
  if course == "python":
    print("Instructor :",details["instructor"])
    print("Duration : ",details["duration"],"hrs")
```

```python
# Name and total sale of each salesperson and highest total sales:
sales = [
    ["aman",12000,15000,18000],
    ["priya",20000,17000,14000],
    ["rohit",16000,22000,19000]
]
highest_sale = 0
top_saleperson=""
for person in sales:
  name=person[0]
  total=sum(person[1:])

  print("Name: ",name)
  print("Total: ",total)
  print("=========================")

  if total>highest_sale:
    highest_sale=total
    top_saleperson=name

print("Top Saleperson: ",top_saleperson)
print("Highest Sale: ",highest_sale)

```

```python
# EXERCISE - 2
def calculate_discount(bill_amount,discount=5):
  discount_amount = (discount/100)*bill_amount
  total_amount = bill_amount - discount_amount
  if bill_amount == 0:
    print("Invalid Bill Amount")
  elif bill_amount >= 1000:
    print("discount amount :", discount_amount)
    print("Total amount :", total_amount)
  elif bill_amount >= 4999:
    print("discount amount :", discount_amount)
    print("Total amount :", total_amount)
  elif bill_amount >= 9999:
    print("discount amount :", discount_amount)
    print("Total amount :", total_amount)
  elif bill_amount <= 10000:
    print("Orignal Bill",bill_amount)
    print("discount amount :", discount_amount)
    print("Total amount :", total_amount)


calculate_discount(10000)
calculate_discount(10000,10)
calculate_discount(10000,15)
calculate_discount(10000,20)




```
