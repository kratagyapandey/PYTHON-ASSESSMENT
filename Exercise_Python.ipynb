```python
# Mirron image
n = 5
for i in range(1,2*n):
  if i <= n:
    star=i
  else:
    star = 2*n-i
  space = n-star
  for j in range(0,space):
    print(" ",end=" ")
  for k in range(0,2*star-1):
    print("* ",end="")

  print()
```

```python
# Display the name and price of headphone
product = [["Watch",3000],["Headphone",2000],["Camera",35000]]

for item in product:
  if item[0]=="Headphone":
    print("Name :",item[0])
    print("Price :",item[1])
```

```python
# [ Instructor name , Duration ] python course
course = {
    "java":{"instructor":"priya","duration":32},
    "sql":{"instructor":"riya","duration":32},
    "python":{"instructor":"Abhisake","duration":42},
    "php":{"instructor":"naman","duration":32}
}
for course,details in course.items():
  if course == "python":
    print("Instructor :",details["instructor"])
    print("Duration : ",details["duration"],"hrs")
```

```python
# Name and total sale of each salesperson and highest total sales:
sales = [
    ["aman",12000,15000,18000],
    ["priya",20000,17000,14000],
    ["rohit",16000,22000,19000]
]
highest_sale = 0
top_saleperson=""
for person in sales:
  name=person[0]
  total=sum(person[1:])

  print("Name: ",name)
  print("Total: ",total)
  print("=========================")

  if total>highest_sale:
    highest_sale=total
    top_saleperson=name

print("Top Saleperson: ",top_saleperson)
print("Highest Sale: ",highest_sale)

```
